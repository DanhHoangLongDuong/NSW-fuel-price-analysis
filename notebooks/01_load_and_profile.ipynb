```python
folder_path = Path("../data/raw")

merge_file = []

for file_path in sorted(folder_path.iterdir()):
    if file_path.suffix.lower() == ".csv":
        merge_file.append(pd.read_csv(file_path, dtype=str, keep_default_na=False).assign(source_file=file_path.name))
    elif file_path.suffix.lower() == ".xlsx":
        merge_file.append(pd.read_excel(file_path, dtype=str, keep_default_na=False).assign(source_file=file_path.name))

full_table = pd.concat(merge_file, ignore_index=True)
full_table.head()
```

### Completeness

```python
# Make a table to check missing, blank and placeholder values inside each column in the table
blank_space = []
for col in full_table.columns:
    if col == "source_file":
        continue
        
    column = col
    missing = full_table[col].isna().sum()
    blank = full_table[col].str.strip().eq("").sum()
    placeholder = full_table[col].str.strip().str.lower().isin(["-", "n/a", "null", "?", ".", "none", "nan", "na", "unknown"]).sum()
    symbol_only = full_table[col].str.contains(r"^[^a-zA-Z0-9]+$", regex=True).sum()
    changed_by_trim = (full_table[col] != full_table[col].str.strip()).sum()
    dictionary = {"column": column,
                  "missing": missing,
                  "blank": blank,
                  "placeholder":placeholder,
                  "symbol_only": symbol_only,
                  "changed_by_trim": changed_by_trim}
    
    blank_space.append(dictionary)
    
summary_table = pd.DataFrame(blank_space).set_index("column")
```

#### Summary

- Method: 5 checks per column (missing, blank after trimming, known placeholder after lowecasing, symbol_only values, and changed_by_trim).
- Result: no missing, blank, placeholder, symbol_only in any of the 8 columns.
- Finding: 2 `Address` rows with whitespace.
- Conclusion: Data is complete. No rows need removing for missing values

#### PriceUpdatedDate

```python
# Replace digits with '#' to expose the format structure
formats = (full_table["PriceUpdatedDate"]
           .astype(str)
           .str.replace(r'[a-zA-Z]', 'w', regex=True)
           .str.replace(r'\d', '#', regex=True))
formats.unique()
```

```python
pd.crosstab(full_table["source_file"], formats, margins=True)
```

```python
# Make 5 different into 2 groups: ISO (containing `-`) and slash (containing `/`)
formats_group = {"####-##-## ##:##:##":"ISO", '#/##/#### #:##': "slash", '#/##/#### ##:##': "slash", '##/##/#### #:##': "slash", '##/##/#### ##:##': "slash"}
formats_table = pd.crosstab(full_table["source_file"], formats.map(formats_group), margins=True)
formats_table
```

```python
# second number of "PriceUpdatedDate" in October 2025 is month
oct2025 = (full_table
                .loc[(full_table["source_file"] == "price_history_checks_oct2025.csv"),["source_file", "PriceUpdatedDate"]]
                .assign(month = lambda x: x["PriceUpdatedDate"].str.split("/").str[1])
          )
oct2025["month"].unique()
```

```python
# first number go above 12 so the first number is day
oct2025 = (full_table
                .loc[(full_table["source_file"] == "price_history_checks_oct2025.csv"),["source_file", "PriceUpdatedDate"]]
                .assign(day = lambda x: x["PriceUpdatedDate"].str.split("/").str[0])
          )
oct2025["day"].unique()
```

Summary:
- 2 Formats: ISO (1189200 lines in 12 files) and slash (69132 lines in only 1 file: `price_history_checks_oct2025.csv`)
- Not a CSV and Excel files difference: other CSV files except `price_history_checks_oct2025.csv` use ISO format
- Slash format is day first: day/month/year
- Slash format have no seconds
- The risk: Automatic parsing reads slash format as month first, which can cause it gives the wrong dates.
- The proposed fix: parse with a stated format for each source

#### Postcode

```python
# Extract state from address
non_NSW_stations_check1 = (full_table["Address"]
                           .str.upper()
                           .str.findall(r"\b(NSW|ACT|VIC|QLD|SA|WA|TAS|NT|"
                                        r"NEW\s+SOUTH\s+WALES|AUSTRALIAN\s+CAPITAL\s+TERRITORY|VICTORIA|QUEENSLAND|SOUTH\s+AUSTRALIA|WESTERN\s+AUSTRALIA|TASMANIA|NORTHERN\s+TERRITORY)\b"
                                        ).str[-1].replace({"NEW SOUTH WALES": "NSW", 
                                                           "AUSTRALIAN CAPITAL TERRITORY": "ACT",
                                                           "VICTORIA":"VIC",
                                                           "QUEENSLAND": "QLD",
                                                           "SOUTH AUSTRALIA": "SA",
                                                           "WESTERN AUSTRALIA": "WA",
                                                           "TASMANIA": "TAS",
                                                           "NORTHERN TERRITORY": "NT"})
                           )
non_NSW_stations_check1.value_counts()
```

```python
# Confirm if all Station are in NSW
non_NSW_stations_check2 = (
                            (full_table["Postcode"].astype("int16").between(1000, 1999)) | 
                            (full_table["Postcode"].astype("int16").between(2000, 2599)) | 
                            (full_table["Postcode"].astype("int16").between(2619, 2899)) | 
                            (full_table["Postcode"].astype("int16").between(2921, 2999))
                          ).map({True:"NSW", False:"Not NSW"})

non_NSW_stations_check2
```
